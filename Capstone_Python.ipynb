Import required libraries

```python
import pandas as pd
import numpy as np
```

# Task 1 : Create three Dataframes and save them as CSV Files

```python
# Project DataFrame
project_df = pd.DataFrame({
    "ID": ["A001", "A002", "A003", "A004", "A005", "A002", "A005", "A003", "A001", "A003", "A001", "A004", "A004", "A005"],
    "Project": ["Project 1", "Project 2", "Project 3", "Project 4", "Project 5", "Project 6", "Project 7", "Project 8", "Project 9", "Project 10", "Project 11", "Project 12", "Project 13", "Project 14"],
    "Cost": [1002000, 2000000, 4500000, 5500000, None, 680000, 400000, 350000, None, 300000, 2000000, 1000000, 3000000, 200000],
    "Status": ["Finished", "Ongoing", "Finished", "Ongoing", "Finished", "Failed", "Finished", "Failed", "Ongoing", "Finished", "Failed", "Ongoing", "Finished", "Finished"]
})

# Employee DataFrame
employee_df = pd.DataFrame({
    "ID": ["A001", "A002", "A003", "A004", "A005"],
    "Name": ["John Alter", "Alice Luxumberg", "Tom Sabestine", "Nina Adgra", "Amy Johny"],
    "Gender": ["M", "F", "M", "F", "F"],
    "City": ["Paris", "London", "Berlin", "Newyork", "Madrid"],
    "Age": [25, 27, 29, 31, 30]
})

# Seniority Level DataFrame
seniority_df = pd.DataFrame({
    "ID": ["A001", "A002", "A003", "A004", "A005"],
    "Designation Level": [2, 2, 3, 2, 3]
})

# Save to CSV
project_df.to_csv("project.csv", index=False)
employee_df.to_csv("employee.csv", index=False)
seniority_df.to_csv("seniority_level.csv", index=False)
```

# Task 2 : Replace missing cost values with running average using a for loop 

```python
project = pd.read_csv("project.csv")

running_sum = 0
running_count = 0
filled_costs = []

for value in project["Cost"]:
    if pd.isna(value):
        if running_count == 0:
            filled_costs.append(np.nan)
        else: 
            filled_costs.append(running_sum / running_count)
    else: 
        filled_costs.append(value)
        running_sum += value
        running_count += 1

project["Cost"] = filled_costs
print(project)
        
```

# Task 3: Split Name into First Name and Last Name, drop Name

```python
employee = pd.read_csv("employee.csv")

employee[["First Name", "Last Name"]] = employee ["Name"].str.split(" ", n=1, expand=True)
employee = employee.drop(columns=["Name"])
print(employee)
```

# Task 4: Join all three DataFrames into one Called Final

```python
seniority = pd.read_csv ("seniority_level.csv")

final = project.merge(employee, on="ID").merge(seniority, on="ID")
print(final)
```

# Task 5: Add Bonus Column - 5% of project cost for Finished Projects 

```python
final["Bonus"] = final.apply(
    lambda row: 0.05 * row["Cost"] if row ["Status"] == "Finished" else 0, axis=1
)
print(final[["ID", "Project", "Cost", "Status", "Bonus"]])
```

# Task 6: Demote Designation level by 1 for Failed projects, delete records with designation >4

```python
for idx in range (len(final)):
    if final.loc[idx, "Status"] == "Failed": 
        final.loc[idx, "Designation Level"] -= 1

final = final[final["Designation Level"]<= 4].reset_index(drop=True)
print(final[["ID", "Project", "Status", "Designation Level"]])
    
```

# Task 7: Add Mr./Mrs. to First Name and drop Gender Column

```python
for idx in range(len(final)):
    if final.loc[idx, "Gender"] == "M":
        final.loc[idx, "First Name"] = "Mr. " + final.loc[idx, "First Name"]
    elif final.loc[idx, "Gender"] == "F":
        final.loc[idx, "First Name"] = "Mrs. " + final.loc[idx, "First Name"]

final = final.drop(columns=["Gender"])
print(final[["ID", "First Name", "Last Name"]])
```

# Task 8: Promote designation level by 1 for employees whose age > 29 using IF 

```python
for idx in range(len(final)):
    if final.loc[idx, "Age"] > 29:
        final.loc[idx, "Designation Level"] += 1

print(final[["ID", "First Name", "Age", "Designation Level"]])
```

# Task 9: Total Project cost per Employee

```python
TotalProjCost = final.groupby(["ID", "First Name"], as_index=False)["Cost"].sum()
TotalProjCost = TotalProjCost.rename(columns={"Cost": "Total cost"})
print(TotalProjCost)
```

# Task 10: Print employee Details whose city contains letter "o"

```python
result_task10 = final[final["City"].str.contains("o", case=False, na=False)]
print(result_task10)
```
